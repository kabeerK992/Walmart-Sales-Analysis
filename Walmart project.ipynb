##DATA CLEANING

```python
import pandas as pd
```

```python
df = pd.read_csv("Walmart.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df['unit_price'] = df['unit_price'].str.replace('$', '').astype(float)
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df['Total_Price'] = df['unit_price'] * df['quantity']
```

```python
df.head()
```

```python
df.describe()
```

```python
df.column = df.columns.str.lower()
```

```python
df.to_csv('clean_walmart.csv', index = False)
```

```python
from sqlalchemy import create_engine


engine = create_engine("postgresql+psycopg2://Username:Password@localhost:5432/walmart_db")

# Upload dataframe to PostgreSQL
df.to_sql(
    name="walmart",   # Table name in pgAdmin
    con=engine,
    if_exists="replace",     # Options: 'replace', 'append'
    index=False
)

print("✅ Data successfully transferred to PostgreSQL!")
```
